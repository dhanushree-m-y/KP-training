```python
class Vehicle:
    def move_right(self):
        print("Vehicle is moving right")
    def move_left(self):
        print("Vehicle is moving left")
    def move_up(self):
        print("Vehicle is moving up")
    def move_down(self):
        print("Vehicle is moving down")

class Car(Vehicle):
    def drive(self):
        print("Car is driving")

V=Vehicle()
V.move_right()

C=Car()
C.drive()
C.move_up()
```

```python
#sigle inheritance
#here bike inherits the properties from the vehicle
class Vehicle:
    def move(self):
        print("Vehicle is moving")
    def drive(self):
        print("Vehicle is driving")
    def stop(self):
        print("Vehicle is stopping")
class Bike(Vehicle):
    def drive_bike(self):
        print("Bike is driving")
B=Bike()
B.drive()
B.move()
B.drive_bike()
```

```python
#multilevel inheritnace
class Vehicle:
    def move(self):
        print("Vehicle is moving")
    def drive(self):
        print("Vehicle is driving")
    def stop(self):
        print("Vehicle is stopping")
class Bike(Vehicle):
    def drive_bike(self):
        print("Bike is driving at 50km/hr")
class Superbike(Bike):
    def drive_super_bike(self):
        print("Bike is driving at 150km/hr at maximum")

B=Bike()
B.drive()
B.move()
B.drive_bike()


S=Superbike()
S.drive_bike()
S.move()
S.drive_bike()


```

```python
#multiple inheritance

class Color:
    def info(self):
        print("Color is colorful")

class Red:
    def info_red(self):
        print("Red is red")
class Black:
    def info_black(self):
        print("Black is black")

class mixed_color(Red, Black,Color):
    def info_mixed_color(self):
        print("Mixed color is from red and black")

R=Red()
B=Black()
M=mixed_color()


M.info()
```
