```python
#Write a function reverse_list(lst) that returns the reversed list without using slicing.

def reverse_list(lst):
    lst.reverse()
    return lst
new_list=[1,2,3,4,5]
new_list.reverse()
print(new_list)
```

```python
#Print all prime numbers between 1 and 50.

for num in range(2, 51):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```
