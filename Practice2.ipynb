```python
products = []
cart = []

def add_product(product_name, price):
    products.append({"name": product_name, "price": price})

def view_products():
    for product in products:
        print(product["name"], product["price"])

def add_to_cart(product_name, quantity):
    cart.append({"name": product_name, "quantity": quantity})

def print_bill():
    total = 0
    for item in cart:
        for product in products:
            if product["name"] == item["name"]:
                amount = product["price"] * item["quantity"]
                print(item["name"], "X", item["quantity"], "=", amount)
                total += amount
    print("Total:", total)

# Example usage
add_product("banana", 23)
view_products()
add_to_cart("banana", 10)
print_bill()
```
